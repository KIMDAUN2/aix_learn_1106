# 선형회귀- 다중회귀(Multiple Regression)
- 여러 입력 특성을 이용해 하나의 값을 예측
- (길이,높이,두께) 특성이 여러개 일 때
- 예측 무게 = a × 길이 + b × 높이 + c × 두께 + d

```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression 
from sklearn.preprocessing import PolynomialFeatures    # 다항특성 자동생성 라이브러리
from sklearn.metrics import mean_absolute_error   # 평균절대값오차 오차범위
from sklearn.preprocessing import StandardScaler  # 정규화클래스
from sklearn.model_selection import train_test_split
```

```python
#데이터
df =pd.read_csv('data/perch_full.csv')
df.head(2)
```

```python
# 무게 데이터,정답
perch_weight = [5.9, 32.0, 40.0, 51.5, 70.0, 100.0, 78.0, 80.0, 85.0, 85.0,
    110.0, 115.0, 125.0, 130.0, 120.0, 120.0, 130.0, 135.0, 110.0,
    130.0, 150.0, 145.0, 150.0, 170.0, 225.0, 145.0, 188.0, 180.0,
    197.0, 218.0, 300.0, 260.0, 265.0, 250.0, 250.0, 300.0, 320.0,
    514.0, 556.0, 840.0, 685.0, 700.0, 700.0, 690.0, 900.0, 650.0,
    820.0, 850.0, 900.0, 1015.0, 820.0, 1100.0, 1000.0, 1100.0,
    1000.0, 1000.0]
```

```python
# 개수 확인
len(perch_weight)
```

```python
# 학습데이터,테스트 데이터 분리
train_data,test_data,train_label,test_label = train_test_split(df, perch_weight)
```

```python
train_data[:2]
```

# degree 
-  다항 특성을 만들 때 사용할 최대 차수
ex)[[2,3,4,6,9]]

degree=1  -> 길이,높이

degree=2  ->위 항(2,3) + 길이2(4), 길이x 높이(6), 높이2(9)

degree=3  ->위 항+ 길이3,길이2 x 높이, 길이 x 높이2,높이3

# include_bias
- 항상 값이 1인 특성을 추가할지 정하는 옵션
- 대부분 False

```python
#특성값을 추가
#length,height,width
#length**2,height**2,width**2
from sklearn.preprocessing import PolynomialFeatures 
poly = PolynomialFeatures(degree=2,include_bias=False)
```

```python
# #다항특성추가-PolynomialFeatures

# poly.fit([[2,3]])
# train_poly = poly.transform([[2,3]])
# train_poly
```

```python
#transeform : 정해진 규칙에 따라 데이터를 변환하는 메서드
poly.fit(train_data)
train_poly = poly.transform(train_data)
test_poly = poly.transform(test_data)
print(train_poly.shape)   #(42,9)
```

```python
# 다항식을 만든 형태 확인 가능
print(poly.get_feature_names_out())
# [  34.5   ,   10.26  ,    6.39  , 1190.25  ,  353.97  ,  220.455 , 105.2676,   65.5614,   40.8321]
# [    a,         b,           c,      a2,         ab,        ac,        b2,        bc,        c2]
```

```python
# 데이터 학습
lr = LinearRegression()
lr.fit(train_data,train_label)
# 데이터 예측
predict = lr.predict(test_data)
print('예측: ',predict)
# 정확도
score = lr.score(train_data,train_label)
score2 = lr.score(test_data,test_label)
print('train정확도: ',score)
print('test정확도: ',score2)
```

```python
# degree = 2 /특성을 9개로 변경해서 예측
# 데이터 학습
lr = LinearRegression()
lr.fit(train_poly,train_label)
# 데이터 예측
predict = lr.predict(test_poly)
print('예측: ',predict)
# 정확도
score = lr.score(train_poly,train_label)
score2 = lr.score(test_poly,test_label)
print('train정확도: ',score)
print('test정확도: ',score2)
```
