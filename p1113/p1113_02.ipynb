# 경사하강법
- SGDClassifier : 샘플 하나를 보고 가중치를 수정하는 방식
- SGDClassifier(loss = 'log_loss') : 로그 손실을 기준으로 ,확률적 경사 하강법을 사용해 학습하는 분류 모델

```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso,LogisticRegression,SGDClassifier
from sklearn.metrics import mean_absolute_error   # 평균절대값오차 오차범위
from sklearn.preprocessing import PolynomialFeatures,StandardScaler 
from sklearn.model_selection import train_test_split 
from scipy.special import expit,softmax  #expit:이진분류에 쓰는 로지스틱 함수/ #softmax: 다중분류에 쓰는 로지스틱 함수
```

```python
# 데이터 연결
fish = pd.read_csv('data/fish_logistic.csv')
```

```python
# 데이터 분리
data = fish[fish.columns[1:]].values
label = fish['Species'].values
train_data,test_data,train_label,test_label = train_test_split(data,label,random_state=42)
```

```python
# 정규화
ss = StandardScaler()
ss.fit(train_data)
train_scaled = ss.transform(train_data)
test_scaled = ss.transform(test_data)
```

```python
print(train_scaled.shape)
```

```python
# 데이터 학습
# lr = LogisticRegression()
# loss = 'log_loss' : LogisticRegression log 함수
# max_iter = 10 : 10번 반복
# random_state = 42 : 랜덤선택을 고정 -  연습할 때 랜덤으로 변경하면 정확도도 함께 변경이 되어서 고정시킴
# verbose = 1 : 반복하면서 정보를 보여줌. / 설정하지 않으면 보여주지 않음.
sc= SGDClassifier(loss = 'log_loss',max_iter=10,random_state=42, verbose=0)
sc.fit(train_scaled,train_label)
print("train 정확도: " ,sc.score(train_scaled,train_label))
print("test 정확도: " ,sc.score(test_scaled,test_label))
```

```python
# 동일한 훈련을 반복
# fit : 모든 기울기,절편을 리셋해서 처음부터 시작
# partial_fit : 현재 있는 정보에서 시작
sc.partial_fit(train_scaled,train_label)
print("train 정확도: " ,sc.score(train_scaled,train_label))
print("test 정확도: " ,sc.score(test_scaled,test_label))
```
