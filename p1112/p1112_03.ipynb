```python
#필요한 도구 가져오기
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.metrics import mean_absolute_error   # 평균절대값오차 오차범위
from sklearn.preprocessing import PolynomialFeatures,StandardScaler 
from sklearn.model_selection import train_test_split
```

```python
# 2개만 분류 - 빙어/도미
# KNeighborsClassifier - 다중분류: 특성값이 여러개인것을 가지고 하나의 값을 예측
# 데이터 불러오기
fish = pd.read_csv('data/fish_logistic.csv')
fish.head(2)
```

```python
#fish.info()               #-데이터 구조 확인(행 개수,열 이름, 비어있지 않은 값 개수, 자료형)
#fish.discribe()           #-숫자데이터의 통계확인(평균,표준편차,최솟값 등)
```

```python
fish.columns
```

```python
#데이터 전처리
#data = fish[fish.colunms[1:]]
data = fish[['Weight', 'Length', 'Diagonal', 'Height', 'Width']].values
label = fish['Species'].values

#데이터 분류
train_data,test_data,train_label,test_label = train_test_split(data,label,random_state=42)
train_data[:5]

#정규화 작업
ss = StandardScaler()
ss.fit(train_data)
train_scaled = ss.transform(train_data)
test_scaled = ss.transform(test_data)

# -KNeighborsClassifie: 분류/  -KNeighborsRegressor: 회귀
#데이터학습 - KNN 모델
clf =KNeighborsClassifier(n_neighbors=5)     # n_neighbors: 가장 가까운 이웃 데이터 보기위함
clf.fit(train_scaled,train_label)        
# print('결과 항목 : ',clf.classes_)            # classes_ : 분류모델이 학습한 정답 종류를 확인하는 속성

#데이터예측
predict =clf.predict(test_scaled)
# print('예측생선 : ',predict)

#정확도
print("train정확도:" , clf.score(train_scaled,train_label))
print("test정확도:" , clf.score(test_scaled,test_label))
```

```python
# predict_proba() : 각 종류로 예측할 확률을 보여주는 함수
proba = clf.predict_proba(test_scaled[:5])  
print('결과 항목 : ',clf.classes_)    # label특성 항목 - 알파벳 순서로 정렬
print(np.round(proba,decimals=4))    #decimals : 소수점 아래 몇 자리까지 반올림 할지 지정하는 옵션
```
