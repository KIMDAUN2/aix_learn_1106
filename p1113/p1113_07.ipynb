검증세트
- 검증세트 - 훈련세트 60% 검증세트 20% 테스트세트 20% (테스트세트에 맞추다 보니, 실제 서비스에서 안 맞는 문제 발생)

```python
#데이터 불러오기
wine =pd.read_csv('data/wine.csv')
wine.head(2)
data = wine[wine.columns[:-1]].values
label = wine['class'].values
data[:5]

#데이터분리
train_data,test_data,train_label,test_label = \
    train_test_split(data,label,test_size=0.2,random_state=42)           #test_size : 훈련 데이터 80%, 테스트데이터:20%
```

```python
sub_data,sub2_data,sub_label,sub2_label =\
    train_test_split(train_data,train_label,test_size=0.2, random_state=42)
```

```python
#특성값을 랜덤으로 설정해서 Tree구조
# max_depth=3 :과대적합일떄 max_depth를 줄이면 ,과대적합 해소
dt = DecisionTreeClassifier(max_depth=3, random_state=42)
dt.fit(train_data,train_label)
print()
```

```python
print('sub 정확도 :',dt.score(sub_data,sub_label))
print('sub2 정확도 :',dt.score(sub2_data,sub2_label))
print('test 정확도 :',dt.score(test_data,test_label))
```
