```python
#필요한 도구 가져오기
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False

import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression 
from sklearn.preprocessing import PolynomialFeatures    # 다항특성 자동생성 라이브러리
from sklearn.linear_model import Ridge
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error   # 평균절대값오차 오차범위
from sklearn.preprocessing import StandardScaler  # 정규화클래스
from sklearn.model_selection import train_test_split
```

```python
df= pd.read_csv('data/MultipleLinearRegressionData.csv')
df.head()
df[['hour','absent']]
data = df.iloc[:,:-2].values
label = df.iloc[:,-1].values
label
```

```python
df['place'].value_counts()
```

```python
# 원핫인코딩 작업
from sklearn.preprocessing import OneHotEncoder
place = df['place'].to_numpy().reshape(-1,1)     # 2차원 배열 변경
encoder = OneHotEncoder(sparse_output=False,drop='first')
result = encoder.fit_transform(place)
#data + result 합치기
total_data = np.concatenate((data,result),axis=1)
# 데이터 분리
train_data, test_data, train_label, test_label = train_test_split(total_data, label, random_state=42)
```

```python
lr= LinearRegression()
lr.fit(train_data,train_label)
score= lr.score(train_data,train_label)
score2 =lr.score(test_data,test_label)
print(score)
print(score2)
```

```python
predict = lr.predict(test_data)
print("예측점수: ",predict)
```

```python
# home = 1. , 0.
# Library = 0., 1.
# Cafe = 0., 0.
in_data = input('공부시간,결석횟수,장소를 입력하세요. 예)4.2 , 1 ,Cafe')
in_data = in_data.split(',')
out_data = []
out_data.append(float(in_data[0]))
out_data.append(float(in_data[1]))
if in_data[2] == 'Home':
    out_data.append(1.0)
    out_data.append(0.0)
elif in_data[2] == 'Library':
    out_data.append(0.0)
    out_data.append(1.0)
elif in_data[2] == 'Cafe':
    out_data.append(0.0)
    out_data.append(0.0)

result = lr.predict([out_data])
print(f"1개 예측점수 :{result[0]:.2f} 점")
```
