cross_validate -> cv변수로 전달가능 : 분할기(splitter)
- 1. 분류모델: StratifiedKFold() /회귀모델(예측): KFold()
- 2. n_splits: 분할개수, shuffle : 랜덤으로 섞어서 테스트


```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['font.family'] = 'Malgun Gothic'
matplotlib.rcParams['axes.unicode_minus']=False
import pandas as pd
import numpy as np
from sklearn import svm,metrics
from sklearn.neighbors import KNeighborsClassifier,KNeighborsRegressor
from sklearn.linear_model import LinearRegression,Ridge,Lasso,LogisticRegression 
from sklearn.metrics import mean_absolute_error   # 평균절대값오차 오차범위
from sklearn.preprocessing import PolynomialFeatures,StandardScaler  
from sklearn.model_selection import train_test_split
from scipy.special import expit, softmax    # expit:이진분류,softmax:다중분류 0-1 숫자로 변환
from sklearn.linear_model import SGDClassifier
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.model_selection import cross_validate,StratifiedKFold
```

```python
#데이터 불러오기
wine =pd.read_csv('data/wine.csv')
wine.head(2)
data = wine[wine.columns[:-1]].values
label = wine['class'].values
data[:5]

#데이터분리
train_data,test_data,train_label,test_label = \
    train_test_split(data,label,test_size=0.2,random_state=42)           #test_size : 훈련 데이터 80%, 테스트데이터:20%
```

```python
# default: test score만 출력, return_train_score=True : train score,test score 모두출력
# fit_time: 훈련시간, score-time:score생성하는 시간, test_score:test정확도, train_score:train정확도
# 분류모델: StratifiedKFold()
# 회귀모델(예측): KFold()
dt =DecisionTreeClassifier(max_depth=3, random_state=42)
# n_splits: 분할개수, shuffle : 랜덤으로 섞어서 테스트
splitter1 = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
score = cross_validate(dt,train_data,train_label,cv=splitter1)
score
```

```python
print('test 평균정확도 : ',np.mean(score['test_score']))
```
